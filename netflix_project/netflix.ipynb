```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data/netflix_titles.csv")

df.head()
```

```python
df['type'].value_counts().plot(kind='pie')
plt.title("Movies vs TV Shows")
plt.show()
plt.figure(figsize=(10,6))
```

```python
df['country'].value_counts().head(10).plot(kind='bar')
plt.title("top 10 countries with most content")
plt.show()
```

```python
df['director'].value_counts().head(40).plot(kind="bar")
plt.title("top 40 directors with most content")
plt.show()
plt.figure(figsize=(10,6))
```

```python
df['duration'].value_counts().head(10).plot(kind='bar') 
plt.title("top 10 durations with successful movies/tv shows")
plt.show()
```

```python
df['director'].value_counts().head(30).plot(kind="pie")
plt.title("top 30 directors")
plt.show()
```
